def years_to_double(rate_pct):
    capacity, years = 100, 0
    while capacity < 200:
        capacity *= 1 + rate_pct / 100
        years += 1
    return years

def first_over(readings, limit):

    counts = {}
    for c in codes:
        counts[c] = counts.get(c, 0) + 1

def old_assets(ages, min_age):

import math, statistics

def spread(values):
    return round(statistics.median(values), 2), round(statistics.stdev(values), 2)

def rms(values):
    return round(math.sqrt(sum(v * v for v in values) / len(values)), 3)

import csv

def read_temps(path):
    temps = {}
    with open(path) as f:
        for row in csv.DictReader(f):
            lst = temps.setdefault(row["asset_id"], [])
            try:
                lst.append(float(row["temp_c"]))
            except ValueError:
                continue
    return temps
